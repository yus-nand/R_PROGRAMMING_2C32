## Sequences of Dates and Alphabets


### Sequence of first day of years

```r
seq(
  as.Date("2010-01-01"),
  as.Date("2017-01-01"),
  by = "years"
)
```

### Sequence of days

```r
seq(
  as.Date("2017-01-01"),
  by = "days",
  length.out = 6
)
```

### Sequence of months

```r
seq(
  as.Date("2017-01-01"),
  by = "months",
  length.out = 6
)
```

### Sequence by years

```r
seq(
  as.Date("2017-01-01"),
  by = "years",
  length.out = 6
)
```

### Sequence using defined start and end dates

```r
startdate <- as.Date("2016-1-1")
enddate <- as.Date("2017-1-1")

out <- seq(enddate, startdate, by = "-1 month")
out
```

## Lecture 28 — Repeats
Source examples: slides/pages 29–42 of the supplied PDF.

### `rep()` syntax

```r
n=7
rep(x, times = n)

rep(x, each = n)

rep(x, length.out = n)
rep(x, length = n)
rep_len(x, length.out = n)
```

### Help for `rep()`

```r
help("rep")
```

### Repeat an object n times

```r
rep(3.5, times = 10)
rep(1:4, 2)
```

### `times` versus `each`

```r
x <- 1:4
x

rep(x, times = 3)
rep(x, each = 3)
```

### Combine `each` and `times`

```r
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
```

### Repeat every object a different number of times

```r
rep(1:4, 2:5)
```

### Generate repetition counts with `seq()`

```r
ans <- seq(from = 2, to = 8, by = 2)
ans

rep(1:4, ans)
```

### Repeating a matrix

```r
x <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x
rep(x, 3)
```

### Repetition of character vectors

```r
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)
```

### Repetition for a pre-specified length

```r
rep(2, length.out = 5)
rep(2, length = 5)

rep(c(2, 3), length = 5)
rep(c(2, 3, 4), length = 5)

rep("apple", length = 5)
rep(c("a", "b", "c"), length = 2)
rep(c("a", "b", "c"), length = 5)
```

## Lecture 29 — Sorting, Ordering and Mode
Source examples: slides/pages 44–55 of the supplied PDF.

### Sorting with `sort()`

```r
y <- c(8, 5, 7, 6)
y

sort(y)
sort(y, decreasing = TRUE)
```

### Ordering with `order()`

```r
y <- c(9, 8, 5, 7, 6)
y

# Positions that produce ascending order
order(y)

# Positions that produce descending order
order(y, decreasing = TRUE)
```

### Mode of R objects

```r
mode(2.432)
mode(c(3, 4, 5, 6, 7, 8))
mode("India")
mode(c("India", "CANADA"))

mode(factor(c("UP", "MP")))
mode(list("India", "USA"))
mode(data.frame(x = 1:2, y = c("India", "USA")))
mode(print)
```

## Lecture 30 — Lists
Source examples: slides/pages 57–68 of the supplied PDF.

### Matrices used in the list examples

```r
x1 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x2 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 5:8,
  byrow = TRUE
)

x1
x2
x1 + x2
```

### Changing an element and observing the type coercion

```r
x1[2, 1] <- "hello"
x1

x1 + x2
```

### Lists can contain matrices

```r
x1 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x2 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 5:8,
  byrow = TRUE
)

matlist <- list(x1, x2)
matlist
```

### Access elements of a list by position

```r
matlist[1]
matlist[2]
```

### A list containing different object types

```r
z1 <- list(
  c("water", "juice", "lemonade"),
  rep(1:4, each = 2),
  matrix(data = 5:8, nrow = 2, ncol = 2, byrow = TRUE)
)

z1
```

### Extracting list elements with `[[ ]]`

```r
z1[[1]]
```

### Difference between `[ ]` and `[[ ]]`

```r
z1[1][2]


z1[[1]][2]
```
